# While Loops and More Loops

A while loop is similar to the if and for loops because it also iterates over a variable. The structure of a while loop is similar to for and if loops:

while "condition" is true:

    Do Something

Lets make a while loop that prints "yeah!" as long as a variable is less than 6.

Now let's nest an if statement inside to break out of the loop when our variable is equal to 3. The break statement forces the loop to end.

Next we will combine our while loop with an else statement. The else statement will only run once the while loop condition becomes false.

Be careful though! You can accidentally create an infinite while loop -- always pay attention to what condition will stop the loop.

#### Continue Statement

The continue statement can be used inside of a while loop. The continue statement skips the current iteration.

In the following example let's define a variable and set it equal to "hippopotomonstrosesquippedaliophobia". If we want to remove all instances of the letter "i" from the word, but print every other letter we can nest an if statement inside of our while loop. 

If the letter is equal to "i" we will skip it using continue.

#### Pass Statement

Finally the pass statement can be used inside of a loop. The pass statement does nothing at all, it is usually used as a placeholder.
